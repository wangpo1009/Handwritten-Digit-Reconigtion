# CNN (CONVOLUTIONAL NEURAL NETWORK)


# 1. Vì sao sử dụng CNN
- Đa phần các điểm liền kề trên hình ảnh đều có quan hệ với nhau, khi biến hình ảnh thành vector số thông thường sẽ làm mất thông tin về quan hệ này => **CNN dùng Convolution để cải thiện**

# 2. Cấu trúc CNN
[![Introduction to Convolution Neural Network - GeeksforGeeks|598](https://media.geeksforgeeks.org/wp-content/uploads/20260217113409515144/working_of_cnn__.webp)
- Input layer: Dữ liệu đầu vào
- Convolution Layer: Mỗi neural là một convolution giữa Receptive field với filter:
		- Filter: Như trọng số, được set kích thước trước và sẽ thay đổi trong quá trình học, mỗi filter sẽ học 1 feature khác nhau
![](https://miro.medium.com/v2/resize:fit:818/1*OVuUVMv0qwRu_nhm9wRtcA.gif)
		- Receptive field: Một vùng nhỏ trong input có kích thước bằng filter, CONV sẽ scan Receptive field theo Stride cho trước => ***Học được các quan hệ giữa các điểm kề nhau***

- Activation Function:  Áp dụng các Activation Function (như ReLU, sigmoid,...), thường áp dụng sau các Linear Layer để:
		- Biến quan hệ qua Convolutional layer, FC layer từ tuyến tính thành phi tuyến.
		- Model học được các quan hệ phức tạp (giảm tính phi tuyến), đa dạng hóa data
![circuit design - What are the activation functions that can be generated  using op-amps and filters? - Electrical Engineering Stack Exchange](https://i.sstatic.net/itkXP.png)
- Pooling layer: Giảm kích thước ảnh để giữ lại các điểm quan trọng mà filter quan tâm => **Giảm overfit, tăng tốc độ dự đoán**:
		- Max Pooling: Lấy điểm cao nhất trong một kích thước cục bộ cho trước (2x2, 3x3,...)
		- Average Pooling: Lấy trung bình trong cục bộ

[![Toy example illustrating the drawbacks of max pooling and average... |  Download Scientific Diagram](https://www.researchgate.net/publication/300020038/figure/fig2/AS:404961465782275@1473561745471/Toy-example-illustrating-the-drawbacks-of-max-pooling-and-average-pooling.png)]
- Flatten: Trải phẳng ma trận nhiều chiều để dễ dàng làm việc ở các bước sau
- Fully Connected Layer: Các neuron là tổ hợp tuyến tính của mọi neuron ở các layer trước nó, với W và Bias khác nhau.[![Using Matrix to Represent Fully Connected Layer and Its Gradient | by  FunCry | Medium](https://miro.medium.com/v2/resize:fit:1194/1*7zVhDlr0P0mYpYyLcygixQ.png)](https://www.google.com/url?sa=t&source=web&rct=j&url=https%3A%2F%2Fmedium.com%2F%40funcry%2Fusing-matrix-to-represent-fully-connected-layer-and-its-gradient-35604d99d138&ved=0CBYQjRxqFwoTCPDYopHt-5QDFQAAAAAdAAAAABAg&opi=89978449)
- Output layer: FC layer cuối -> Activation Function -> Xác suất
- Tìm hiểu thêm:
		- Normalization Layer: Chuẩn hóa các giá trị để cải thiện việc train
		- Dropout Layer: Các neuron có thể bị drop trọng số theo xác suất cho trước ở MỖI iteration => Gần giống idea L1

- Stride: Bước nhảy các điểm khi lớp CONV quét input
- Zero-padding: Số lớp border = 0 tạo thêm

## 3.2 Training Loop

```python
from re import X
# Progess bar
from tqdm.auto import tqdm

#Set seed and start timer
torch.manual_seed(36)
start = timer()

epochs = 1

# Training and test loop
for epoch in tqdm(range(epochs)):
  print(f"-----Epoch: {epoch}\n------")

  ##training
  train_loss = 0 #Set back to 0 to avoid manipulating from bathes loop
  # traning batches loop
  for batch, (x, y) in enumerate(train_dataloader):
    model_0.train() #Put model in train mode
    #1. Forward pass
    y_pred = model_0(x)

    #2. Calc loss
    loss = loss_fn(y_pred, y)
    train_loss += loss

    #3. Optimise
    optimizer.zero_grad() #Set gradint to 0 to void gradient manipulating

    #4. Loss Backward
    loss.backward() # Calc new derivative

    #5. Optimizer Step
    optimizer.step() #Take optimising step

    # Print
    if batch % 400 == 0:
      print(f"Looked at {batch * len(x)}/{len(train_dataloader.dataset)} samples")

  train_loss /= len(train_dataloader)

  ##Testing
  test_loss, test_acc = 0, 0
  model_0.eval() #Put model in eval mode
  with torch.inference_mode():
    for X_test, Y_test in test_dataloader:
      #1. forward pass
      test_pred = model_0(X_test)

      #2. Calc loss
      test_loss += loss_fn(test_pred, Y_test).item()

      #3. Calc acc
      test_acc += accuracy_fn(Y_test, test_pred.argmax(dim = 1))

  test_loss /= len(test_dataloader)
  test_acc /= len(test_dataloader)
  print(f"Train loss: {train_loss:.5f} | Test loss: {test_loss:.5f} | Test acc: {test_acc:.5f}")

end = timer()
total_time = print_train_time(start = start, end = end, device = "cpu")
print(f"Total training time: {total_time:.3f} seconds")
```

# 4: CNN

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```

```python
def train_step(model: torch.nn.Module,
               data_loader: torch.utils.data.DataLoader,
               loss_fn: torch.nn.Module,
               optimizer: torch.optim.Optimizer,
               accuracy_fn,
               device: torch.device = device):
    train_loss, train_acc = 0, 0
    model.to(device)
    for batch, (X, y) in enumerate(data_loader):
        # Send data to GPU
        X, y = X.to(device), y.to(device)

        # 1. Forward pass
        y_pred = model(X)

        # 2. Calculate loss
        loss = loss_fn(y_pred, y)
        train_loss += loss
        train_acc += accuracy_fn(y_true=y,
                                 y_pred=y_pred.argmax(dim=1)) # Go from logits -> pred labels

        # 3. Optimizer zero grad
        optimizer.zero_grad()

        # 4. Loss backward
        loss.backward()

        # 5. Optimizer step
        optimizer.step()

    # Calculate loss and accuracy per epoch and print out what's happening
    train_loss /= len(data_loader)
    train_acc /= len(data_loader)
    print(f"Train loss: {train_loss:.5f} | Train accuracy: {train_acc:.2f}%")

def test_step(data_loader: torch.utils.data.DataLoader,
              model: torch.nn.Module,
              loss_fn: torch.nn.Module,
              accuracy_fn,
              device: torch.device = device):
    test_loss, test_acc = 0, 0
    model.to(device)
    model.eval()
    with torch.inference_mode():
        for X, y in data_loader:
            # Send data to GPU
            X, y = X.to(device), y.to(device)

            # 1. Forward pass
            test_pred = model(X)

            # 2. Calculate loss and accuracy
            test_loss += loss_fn(test_pred, y)
            test_acc += accuracy_fn(y_true=y,
                y_pred=test_pred.argmax(dim=1) # Go from logits -> pred labels
            )

        test_loss /= len(data_loader)
        test_acc /= len(data_loader)
        print(f"Test loss: {test_loss:.5f} | Test accuracy: {test_acc:.2f}%\n")

def print_train_time(start: float,
                     end: float,
                     device: torch.device = None):
  total_time = end - start
  print(f"Train time on {device}: {total_time:.3f} seconds")
  return total_time
```

## 4.1: Model

```python
import torch.nn.functional as F
import torch.optim as optim

class CNN(nn.Module):
  def __init__(self):
    super().__init__()
    self.CNN_stack = nn.Sequential(
        nn.Conv2d(1, 10, kernel_size = 5),
        nn.ReLU(),
        nn.MaxPool2d(2),
        nn.Conv2d(10, 20, kernel_size = 5),
        nn.ReLU(),
        nn.MaxPool2d(2),
    )
    self.FC_stack = nn.Sequential(
        nn.Linear(320, 50),
        nn.ReLU(),
        nn.Dropout(0.5),
        nn.Linear(50, 10)
    )
  def forward(self, x):
    x = self.CNN_stack(x)
    x = torch.flatten(x, 1)
    x = self.FC_stack(x)
    return x
```

```python
model_1 = CNN().to(device)
print(model_1)
loss_fn = nn.CrossEntropyLoss()
optimizer = optim.SGD(model_1.parameters(), lr = 0.01, momentum = 0.5)
```

## 4.2: Train

```python
start1 = timer()
epochs = 3
for epoch in tqdm(range(epochs)):
  print(f'-----Epoch: {epoch}\n-----')
  train_step(model = model_1,
             data_loader = train_dataloader,
             loss_fn = loss_fn,
             optimizer = optimizer,
             accuracy_fn = accuracy_fn,
             device = device
             )
  test_step(data_loader = test_dataloader,
            model = model_1,
            loss_fn = loss_fn,
            accuracy_fn = accuracy_fn,
            device = device)
end1 = timer()

train_time_1 = print_train_time(start = start1, end = end1, device = device)
```

```python
def make_predictions(model : torch.nn.Module,
                     data: list,
                     device: torch.device = device):
  pred_probs = []
  model.eval()
  with torch.inference_mode():
    for sample in data:
      sample = torch.unsqueeze(sample, dim = 0).to(device)
      # print(f"sample shape: {sample.shape}\n")
      # Forward pass
      pred_logit = model(sample)

      # To proability
      pred_prob = torch.softmax(pred_logit.squeeze(), dim = 0)

      # Get off gpu for futher calc
      pred_probs.append(pred_prob.cpu())

    #Stack list to tensor
    return torch.stack(pred_probs) #Return prob list
```

```python
def preprocess_mnist_image(img, invert = True, use_mnist_normalize = False):
    """
    img: PIL Image hoặc ảnh từ Gradio ImageEditor.
    invert=True vì ảnh tự vẽ thường là nền trắng nét đen,
    còn MNIST thường là nền đen nét trắng.
    """

    # Nếu ảnh đến từ Gradio ImageEditor
    if isinstance(img, dict):
        img = img.get("composite") or img.get("background")

    if img is None:
        raise ValueError("Chưa có ảnh đầu vào.")

    # Đưa về grayscale
    if img.mode == "RGBA":
        bg = Image.new("RGBA", img.size, (255, 255, 255, 255))
        bg.alpha_composite(img)
        img = bg.convert("L")
    else:
        img = img.convert("L")

    # Đảo màu: nền trắng nét đen -> nền đen nét trắng
    if invert:
        img = ImageOps.invert(img)

    # Crop vùng có nét số
    arr = np.array(img)
    ys, xs = np.where(arr > 30) # Gán tòa độ y và x của từng pixel vào mảng ys và xs

    if len(xs) == 0 or len(ys) == 0:
        raise ValueError("Không thấy nét số trong ảnh.")

    img = img.crop((xs.min(), ys.min(), xs.max() + 1, ys.max() + 1)) # Hàm crop lấy từ hướng trái qua và phải xuống và dừng ngay trước right, top như vòng for i in

    # Resize sao cho số nằm trong vùng khoảng 20x20 rồi đặt giữa canvas 28x28
    img.thumbnail((20, 20), Image.Resampling.LANCZOS)

    canvas = Image.new("L", (28, 28), 0)
    left = (28 - img.width) // 2
    top = (28 - img.height) // 2
    canvas.paste(img, (left, top))

    # Đưa về tensor shape: [batch, channel, height, width] = [1, 1, 28, 28]
    arr = np.array(canvas).astype("float32") / 255.0
    x = torch.from_numpy(arr).unsqueeze(0).unsqueeze(0) #Thêm chiều Batch_size, channel vì CNN cần [Batch_size, channel, height, width]
                                                        # arr.shape = (28,28)

    # Chỉ bật nếu lúc train bạn có dùng Normalize((0.1307,), (0.3081,))
    if use_mnist_normalize:
        x = (x - 0.1307) / 0.3081

    return x

def predict_digit(img):
    x = preprocess_mnist_image(
        img,
        invert=True,
        use_mnist_normalize=False  # đổi thành True nếu train có Normalize MNIST
    ).to(device)

    with torch.no_grad():
        logits = model_1_loaded(x)
        probs = F.softmax(logits, dim=1)[0]

    pred = torch.argmax(probs).item()

    return pred, probs.cpu().numpy()
```

```python
# @title
device = "cuda" if torch.cuda.is_available() else "cpu"
device

def train_step(model: torch.nn.Module,
               data_loader: torch.utils.data.DataLoader,
               loss_fn: torch.nn.Module,
               optimizer: torch.optim.Optimizer,
               accuracy_fn,
               device: torch.device = device):
    train_loss, train_acc = 0, 0
    model.to(device)
    for batch, (X, y) in enumerate(data_loader):
        # Send data to GPU
        X, y = X.to(device), y.to(device)

        # 1. Forward pass
        y_pred = model(X)

        # 2. Calculate loss
        loss = loss_fn(y_pred, y)
        train_loss += loss
        train_acc += accuracy_fn(y_true=y,
                                 y_pred=y_pred.argmax(dim=1)) # Go from logits -> pred labels

        # 3. Optimizer zero grad
        optimizer.zero_grad()

        # 4. Loss backward
        loss.backward()

        # 5. Optimizer step
        optimizer.step()

    # Calculate loss and accuracy per epoch and print out what's happening
    train_loss /= len(data_loader)
    train_acc /= len(data_loader)
    print(f"Train loss: {train_loss:.5f} | Train accuracy: {train_acc:.2f}%")

def test_step(data_loader: torch.utils.data.DataLoader,
              model: torch.nn.Module,
              loss_fn: torch.nn.Module,
              accuracy_fn,
              device: torch.device = device):
    test_loss, test_acc = 0, 0
    model.to(device)
    model.eval()
    with torch.inference_mode():
        for X, y in data_loader:
            # Send data to GPU
            X, y = X.to(device), y.to(device)

            # 1. Forward pass
            test_pred = model(X)

            # 2. Calculate loss and accuracy
            test_loss += loss_fn(test_pred, y)
            test_acc += accuracy_fn(y_true=y,
                y_pred=test_pred.argmax(dim=1) # Go from logits -> pred labels
            )

        test_loss /= len(data_loader)
        test_acc /= len(data_loader)
        print(f"Test loss: {test_loss:.5f} | Test accuracy: {test_acc:.2f}%\n")

def print_train_time(start: float,
                     end: float,
                     device: torch.device = None):
  total_time = end - start
  print(f"Train time on {device}: {total_time:.3f} seconds")
  return total_time
```

```python
def make_predictions(model : torch.nn.Module,
                     data: list,
                     device: torch.device = device):
  pred_probs = []
  model.eval()
  with torch.inference_mode():
    for sample in data:
      sample = torch.unsqueeze(sample, dim = 0).to(device)
      # print(f"sample shape: {sample.shape}\n")
      # Forward pass
      pred_logit = model(sample)

      # To proability
      pred_prob = torch.softmax(pred_logit.squeeze(), dim = 0)

      # Get off gpu for futher calc
      pred_probs.append(pred_prob.cpu())

    #Stack list to tensor
    return torch.stack(pred_probs) #Return prob list
```

```python
import torch
import matplotlib.pyplot as plt

def show_saliency_map(model, sample, device, class_names=None):
    """
    sample: tensor ảnh MNIST shape [1, 28, 28] hoặc [1, 1, 28, 28]
    """

    model.eval()

    if sample.dim() == 3:
        x = sample.unsqueeze(0)  # [1, 1, 28, 28]
    else:
        x = sample.clone()       # [1, 1, 28, 28]

    x = x.to(device)
    x.requires_grad_(True)

    logits = model(x)
    probs = torch.softmax(logits, dim=1)

    pred_class = probs.argmax(dim=1).item()
    pred_prob = probs[0, pred_class].item()

    # Lấy score của class model dự đoán
    score = logits[0, pred_class]

    model.zero_grad()
    score.backward()

    # Gradient theo pixel ảnh
    saliency = x.grad.abs().squeeze().detach().cpu()

    # Ảnh gốc để hiển thị
    img = x.detach().squeeze().cpu()

    plt.figure(figsize=(8, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(img, cmap="gray")
    plt.title("Ảnh gốc")
    plt.axis(False)

    plt.subplot(1, 2, 2)
    plt.imshow(saliency, cmap="hot")
    plt.title(f"Saliency | Pred: {pred_class} ({pred_prob:.2%})")
    plt.axis(False)

    plt.show()

def saliency_on_digit_ratio(model, sample, device, threshold=0.2):
    model.eval()

    if sample.dim() == 3:
        x = sample.unsqueeze(0)
    else:
        x = sample.clone()

    x = x.to(device)
    x.requires_grad_(True)

    logits = model(x)
    pred_class = logits.argmax(dim=1).item()

    score = logits[0, pred_class]

    model.zero_grad()
    score.backward()

    saliency = x.grad.abs().squeeze().detach().cpu()
    img = x.detach().squeeze().cpu()

    # mask vùng có nét số
    digit_mask = img > threshold

    total_saliency = saliency.sum()
    digit_saliency = saliency[digit_mask].sum()

    ratio = digit_saliency / (total_saliency + 1e-8)

    return ratio.item(), pred_class
```
